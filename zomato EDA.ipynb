```python
df_c=pd.read_excel('Country_Code.xlsx')
df_c.head()
```

```python
final_df=pd.merge(df,df_c,on='Country Code',how='left') #merging two data sets on same column country code by joing left
final_df.head(3)
```

```python
final_df.Country.value_counts() #value_counts count the values
```

```python
country_names=final_df.Country.value_counts().index #index shows names ,values shows values only
country_names
```

```python
country_val=final_df.Country.value_counts().values #values shows values only
country_val
```

```python
#pie chart
#plt.pie(country_val,labels=country_names)
#top 3 countries used zomato
plt.pie(country_val[:3],labels=country_names[:3],autopct='%1.2f%%')
```

```python
final_df.columns
```

```python
# do more relationship things

final_df.groupby(['Aggregate rating','Rating color','Rating text']).size()
```

```python
ratings=final_df.groupby(['Aggregate rating','Rating color','Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
```

```python
ratings
```

## Observations:
1. when the ratings is between 4.5 to 4.9 ---> Excellent ,shown in Drak Green
2. when the ratings is between 4.0 to 4.4 ---> Very Good ,shown in  Green
3. when the ratings is between 3.5 to 3.9 ---> Good, shown in Yellow
4. when the ratings is between 2.5 to 3.4 ---> Avarage, shown in Orange
5. when the ratings is between 1.8 to 2.4 ---> Poor shown in Red
6. when the ratings is 0.0 ---> Not Rated shown in White ,many people not rated

```python
ratings.head()
```

```python
plt.figure(figsize=(12,8))
sns.barplot(x="Aggregate rating",y="Rating Count",data=ratings,hue="Rating text")
```

```python
plt.figure(figsize=(12,8))
sns.set_theme(style='darkgrid')
sns.barplot(x="Aggregate rating",y="Rating Count",data=ratings,hue='Rating color',palette=['white','red','orange','yellow','green','darkgreen'])
```

# observation:
    1.Not Rated count is very high 
    2.Maximum no.of rating are between 2.5 to 3.4

```python
sns.set_theme(style='darkgrid')
sns.countplot(x="Rating color",data=ratings,palette=['silver','red','orange','yellow','green','darkgreen'])
```
